# Rank 5 approximation

```python
def NMF_Decomposition(term_matrix, rank_k):
    
    model = NMF(n_components=rank_k, init="random", max_iter=500, random_state=0)
    C = model.fit_transform(term_matrix)
    D = model.components_
    return C, D
```

```python
def key_sentence_index_func(D, step):
    
    norm_list = np.linalg.norm(D[step:, :], axis=0)
    return np.argmax(norm_list)
```

```python
def change_column_arrange(D, step, index):
    
    D[step:, [step, index]] = D[step:, [index, step]]
    return D
```

```python
def householder(DP, step, rank_k):

    def sgn(x):
        return 1 if x >= 0 else -1

    X = DP[step:, step]
    e = np.zeros(rank_k - step)
    e[0] = 1

    V = X + (sgn(X[0]) * (np.linalg.norm(X) * e))
    I2 = np.identity(rank_k - step)
    H = I2- 2 * (np.outer(V, V.T) / np.dot(V.T, V))

    Q = np.identity(rank_k)
    Q[step:, step:] = H
    
    return Q
```

```python
def Next_C_D(Q, DP, C, step, rank_k):

    Q_DP = np.matmul(Q.T, DP)
    C_Q = np.matmul(C, Q)

    I = np.identity(rank_k)
    I[step][step] = Q_DP[step][step]
    T = I

    C_new = np.matmul(C_Q, T)
    D_new = np.matmul(np.linalg.inv(T), Q_DP)
    
    return C_new, D_new
```

```python
def top_terms(C, step):

    enum_C = list(enumerate(C.T[step]))
    sorted_C = sorted(enum_C,key = lambda x: x[1], reverse=True)
    return sorted_C[:2]
```

```python
def get_key_word_sentence_each_step(term_matrix, rank_k):

    C, D = NMF_Decomposition(term_matrix, rank_k)

    for step in range(rank_k):

        key_sentence_index = key_sentence_index_func(D, step)

        print(f"Key sentence: {step}: {text_without_newline[key_sentence_index]}")

        DP = change_column_arrange(D, step, key_sentence_index)

        Q_householder = householder(DP, step, rank_k)

        C, D = Next_C_D(Q_householder, DP, C, step, rank_k)

        index_of_top_terms = list(map(lambda x: x[0], top_terms(C, step)))

        print(list(map(lambda x: feature_names[x], index_of_top_terms)))
```

```python
get_key_word_sentence_each_step(Tfidf_text, 5)
```
